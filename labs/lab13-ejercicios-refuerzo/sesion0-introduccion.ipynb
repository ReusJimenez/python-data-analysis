### Inteligencia Artificial

![IA](https://th.bing.com/th/id/OIP.p65Pe2wSF5hT-nG07FYPfgHaHa?pid=ImgDet&rs=1)

* La inteligencia es la capacidad de resolver problemas.
* La inteligencia puede definirse como la capacidad tanto de entender o comprender como de resolver problemas.
* La inteligencia puede definirse como la capacidad de percibir o inferir información, y retenerla como conocimiento para aplicarlo a comportamientos adaptativos dentro de un entorno o contexto.

![DLML](https://th.bing.com/th/id/OIP.uKHeuplxedo6iZ0mP1XdmwHaFj?pid=ImgDet&rs=1)

* Disciplina científica que se ocupa de crear programas informáticos que ejecutan operaciones comparables a las que realiza la mente humana, como el aprendizaje o el razonamiento lógico.
