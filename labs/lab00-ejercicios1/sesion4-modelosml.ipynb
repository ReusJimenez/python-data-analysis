```python
import pandas as pd
```

```python
clasificacion = pd.read_csv("./Data/ML-MATT-CompetitionQT2021_train.csv", sep = ";")
clasificacion.head()
```

```python
data = clasificacion[['PRBUsageUL', 'PRBUsageDL', 'meanThr_DL',
       'meanThr_UL', 'maxThr_DL', 'maxThr_UL', 'meanUE_DL', 'meanUE_UL',
       'maxUE_DL', 'maxUE_UL', 'maxUE_UL+DL', 'Unusual']]
data.head()
```

```python
x_data = data[['PRBUsageUL', 'PRBUsageDL', 'meanThr_DL',
       'meanThr_UL', 'maxThr_DL', 'maxThr_UL', 'meanUE_DL', 'meanUE_UL',
       'maxUE_DL', 'maxUE_UL', 'maxUE_UL+DL']]
```

```python
y_data = data["Unusual"]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size = 0.2, random_state = 2024)
```

### Arboles de Decision Clasificación

* Tipo de Aprendizaje = Supervisado
* Tipo de Problema a resolver = predictivo categórico
* Parametros = No hay
* Hiperparámetros = criterio de separacion, estrategia de division, profundidad del arbol

![AD](https://i.ytimg.com/vi/NdFZSkWj4Ko/maxresdefault.jpg)

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
arb_dec_cla = DecisionTreeClassifier(criterion = "gini", splitter = "best", max_depth = 15, random_state = 500)
arb_dec_cla.fit(x_train, y_train)
```

```python
y_pred_adc = arb_dec_cla.predict(x_test)
y_pred_adc
```

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score

print(confusion_matrix(y_test, y_pred_adc))
print(accuracy_score(y_test, y_pred_adc))
print(precision_score(y_test, y_pred_adc))
print(recall_score(y_test, y_pred_adc))
print(f1_score(y_test, y_pred_adc))
```

```python
# Algoritmo Naive-Bayes 0.515
# Maquina de Vectores de Soporte 0.734
# K vecinos más cercanos 0.753
# Regresion Logistica 0.734
```

### Determinacion del número de Grupos

![Codo_de_Jambu](https://i.ytimg.com/vi/3NquQS22Q88/hqdefault.jpg)

```python
import plotly.express as px
import numpy as np
from sklearn.cluster import KMeans
```

```python
grupos = []

for i in range(1, 12):
    kmeans = KMeans(n_clusters = i, max_iter = 300)
    kmeans.fit(x_data)
    grupos.append(kmeans.inertia_)

df = pd.DataFrame({"n_cluster": range(1, 12), "calculo": grupos})
px.line(df, x = "n_cluster", y = "calculo")
```

### Analisis de Clusters
