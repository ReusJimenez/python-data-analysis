```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder

# Cargar datos desde el archivo Excel
df = pd.read_excel("Problemas.xlsx", sheet_name="Regla_Asociacion")

# Convertir los datos en formato de lista de transacciones
transactions = df.groupby("id_compra")["item"].apply(list).tolist()

# Aplicar TransactionEncoder para transformar los datos
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_encoded = pd.DataFrame(te_ary, columns=te.columns_)

# Aplicar el algoritmo Apriori con un soporte mínimo de 0.00025
frequent_itemsets = apriori(df_encoded, min_support=0.00025, use_colnames=True)

# Generar reglas de asociación con confianza mínima de 0.6
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
rules = rules[rules["confidence"] >= 0.6]

# Filtrar reglas que coincidan con las opciones dadas
options = [
    {"antecedents": {"rubbing alcohol", "citrus fruit"}, "consequents": {"herbs", "pip fruit", "frozen vegetables"}},
    {"antecedents": {"citrus fruit", "root vegetables", "tropical fruit", "whole milk"}, "consequents": {"other vegetables"}},
    {"antecedents": {"pip fruit", "butter"}, "consequents": {"whole milk"}},
    {"antecedents": {"sweet spreads", "newspapers"}, "consequents": {"soda"}}
]

# Buscar cuál de las opciones tiene mayor lift
max_lift = 0
best_option = None

for option in options:
    match = rules[(rules["antecedents"] == frozenset(option["antecedents"])) & 
                  (rules["consequents"] == frozenset(option["consequents"]))]
    if not match.empty:
        lift_value = match["lift"].values[0]
        if lift_value > max_lift:
            max_lift = lift_value
            best_option = option

# Mostrar la mejor opción
if best_option:
    print("La regla con mayor lift entre las opciones dadas es:")
    print("Antecedente:", best_option["antecedents"])
    print("Consecuente:", best_option["consequents"])
    print("Lift:", max_lift)
else:
    print("Ninguna de las opciones dadas está en los resultados.")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error

# Cargar los datos desde el archivo Excel
datos = pd.read_excel("Problemas.xlsx", sheet_name="Arbol_de_Decision")

# Separar características (X) y variable objetivo (y)
x_data = datos.drop(columns='esp_vida')
y_data = datos['esp_vida']

# Dividir en conjuntos de entrenamiento y prueba
x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.2, random_state=212211128)

# Definir y entrenar Modelo 1
modelo1 = DecisionTreeRegressor(criterion="squared_error", splitter="best", max_depth=9, random_state=12345)
modelo1.fit(x_train, y_train)

# Definir y entrenar Modelo 2
modelo2 = DecisionTreeRegressor(criterion="friedman_mse", splitter="random", max_depth=9, random_state=67890)
modelo2.fit(x_train, y_train)

# Evaluar modelos con el MSE (Error Cuadrático Medio)
mse1 = mean_squared_error(y_test, modelo1.predict(x_test))
mse2 = mean_squared_error(y_test, modelo2.predict(x_test))

# Comparar los resultados
if mse1 < mse2:
    print(f"El Modelo 1 es mejor con un MSE de {mse1:.4f} vs {mse2:.4f} del Modelo 2.")
else:
    print(f"El Modelo 2 es mejor con un MSE de {mse2:.4f} vs {mse1:.4f} del Modelo 1.")
```
