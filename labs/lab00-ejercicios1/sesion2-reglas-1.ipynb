```python
Prediccion Modelo 1 -> 100 imagenes -> 40 clase A y 60 clase B
Prediccion Modelo 2 -> 100 imagenes -> 20 clase A y 80 clase B

Real                -> 100 imagenes -> 25 clase A y 75 clase B
```

```python
# Modelo 2.1
    A     B
A   05    15  20
B   70    10  80

BC  15 - MC 85 

# Modelo 2.2

    A     B
A   15    05  20
B   10    70  80

BC  85 - MC 15 
```

```python
# Matriz de Confusion


Reales        A - B

Predichos A   BC  MC  Total predichos como A
          B   MC  BC  Total predichos como B
    
# BC = Bien clasificados
# MC = Mal Clasificados
```

```python
# Modelo 1
    A     B
A   25    15 
B   10    50 

BC  75 - MC 25 - Total de datos 100

BC/T = 75/100 = 0.75

# Modelo 2

    A     B
A   20    5
B   10    70

BC  90 - MC 15 - Total de datos 105

BC/T = 90/105 = 0.86
```

```python
# TP = true positive = verdaderos positivos
# FP = false positives = falsos positivos = clasifican como positivos pero en realidad son negativos
# TN = true negative = verdaderos negativos
# FN = false negatives = falsos negativos = clasifican como negativos pero en realidad son positivos
```

```python
# Matriz de Confusion


Reales         (+) (-)

Predichos (+)   tp  fp  
          (-)   fn  tn  
```

### Problemas en el modelamiento de los Datos

Los problemas más comunes son el Overfitting (Sobreajuste) y el Underfitting.

### Overfitting

Los datos predichos por el modelo tienen un % de acierto muy elevado con los datos de entrenamiento, pero muy bajos con los datos de prueba (test)

### Underfitting

Los datos predichos por el modelo no logran superar un % umbral por más etapas de entrenamiento que tenga.
