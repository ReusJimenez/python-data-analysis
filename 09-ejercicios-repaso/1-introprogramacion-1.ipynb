```python
print(0)
print(1)
print(2)
print(3)
print(4)
```

```python
for i in range(0, 5): # [0, 1, 2, 3, 4]
    print(i)
```

```python
for i in [0, 1, 2, 3, 4]:
    print(i)
```

```python
for i in range(0, 5):
    i
```

```python
for var_0 in range(0, 4):
    for var_1 in range(0, 5):
        print(var_0 , var_1)
```

```python
for var0 in range(0, 4):
    for var1 in range(0, 5):
        if var0 == var1:
            print(var0 , var1)
```

```python
for j in range(0, 6):
    if j == 3:
        pass
    else:
        print(j)
```

```python
for estudiantes in lista1:
    print(estudiantes)
```
