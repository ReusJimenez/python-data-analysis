#### Caso 1: Clasificación

```python
import pandas as pd
import numpy as np
```

```python
clasificacion = pd.read_csv("./Data/ML-MATT-CompetitionQT2021_train.csv", sep = ";")
clasificacion.head()
```

```python
data = clasificacion[['PRBUsageUL', 'PRBUsageDL', 'meanThr_DL',
       'meanThr_UL', 'maxThr_DL', 'maxThr_UL', 'meanUE_DL', 'meanUE_UL',
       'maxUE_DL', 'maxUE_UL', 'maxUE_UL+DL', 'Unusual']]
data.head()
```

```python
x_data = data[['PRBUsageUL', 'PRBUsageDL', 'meanThr_DL',
       'meanThr_UL', 'maxThr_DL', 'maxThr_UL', 'meanUE_DL', 'meanUE_UL',
       'maxUE_DL', 'maxUE_UL', 'maxUE_UL+DL']]
```

```python
y_data = data["Unusual"]
```

```python
data["Unusual"].value_counts()
```

```python
x_data.shape
```

```python
y_data.shape
```

```python
y_data.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size = 0.2, random_state = 1000)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
x_test.shape[0] / x_data.shape[0]
```

```python
x_train
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
arb_dec_cla = DecisionTreeClassifier(criterion = "gini", splitter = "best", max_depth = 6, random_state = 1000)
arb_dec_cla.fit(x_train, y_train)
```

```python
y_pred_adc = arb_dec_cla.predict(x_test)
y_pred_adc
```

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score

print(confusion_matrix(y_pred_adc, y_test))
print(accuracy_score(y_pred_adc, y_test))
print(precision_score(y_pred_adc, y_test))
print(recall_score(y_pred_adc, y_test))
print(f1_score(y_pred_adc, y_test))
print(roc_auc_score(y_pred_adc, y_test))
```

```python
(5284 + 1270)
```

```python
confusion_matrix(y_test, y_pred_adc).sum().sum()
```

```python
6554/7381
```

```python
arb_dec_cla_2 = DecisionTreeClassifier(criterion = "entropy", splitter = "best", max_depth = 8, random_state = 1000)
arb_dec_cla_2.fit(x_train, y_train)
```

```python
y_pred_adc_2 = arb_dec_cla_2.predict(x_test)
y_pred_adc_2
```

```python
print(confusion_matrix(y_pred_adc_2, y_test))
print(accuracy_score(y_pred_adc_2, y_test))
print(precision_score(y_pred_adc_2, y_test))
print(recall_score(y_pred_adc_2, y_test))
print(f1_score(y_pred_adc_2, y_test))
print(roc_auc_score(y_pred_adc_2, y_test))
```

#### Caso 2: Regresión

```python
regresion = pd.read_csv("./Data/House_Prices_train.csv")
regresion.head()
```

```python
x_data = regresion[['MSSubClass', 'LotArea', 'OverallQual',
       'OverallCond', 'YearBuilt', 'YearRemodAdd', 'BsmtFinSF1',
       'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', '1stFlrSF', '2ndFlrSF',
       'LowQualFinSF', 'GrLivArea', 'BsmtFullBath', 'BsmtHalfBath', 'FullBath',
       'HalfBath', 'BedroomAbvGr', 'KitchenAbvGr', 'TotRmsAbvGrd',
       'Fireplaces', 'GarageCars', 'GarageArea', 'WoodDeckSF',
       'OpenPorchSF', 'EnclosedPorch', '3SsnPorch', 'ScreenPorch', 'PoolArea',
       'MiscVal', 'MoSold', 'YrSold']]
```

```python
y_data = regresion["SalePrice"]
```

```python
x0_train, x0_test, y0_train, y0_test = train_test_split(x_data, y_data, test_size = 0.2, random_state = 1000)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
arb_dec_reg = DecisionTreeRegressor(criterion = "squared_error", splitter = "best", max_depth = 9, random_state = 1000)
arb_dec_reg.fit(x0_train, y0_train)
```

```python
y0_predict_adr = arb_dec_reg.predict(x0_test)
y0_predict_adr
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mean_squared_error(y0_test, y0_predict_adr)
```

```python
arb_dec_reg_2 = DecisionTreeRegressor(criterion = "friedman_mse", splitter = "random", max_depth = 9, random_state = 1000)
arb_dec_reg_2.fit(x0_train, y0_train)
```

```python
y0_predict_adr_2 = arb_dec_reg_2.predict(x0_test)
y0_predict_adr_2
```

```python
mean_squared_error(y0_test, y0_predict_adr_2)
```

```python
mean_squared_error(y0_test, y0_predict_adr) # Primer modelo (mse)
```

```python
mean_squared_error(y0_test, y0_predict_adr_2) # Segundo modelo (friedman mse)
```
