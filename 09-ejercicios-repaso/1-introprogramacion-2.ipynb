### Unión de Archivos

```python
df1 = pd.DataFrame({'Frutas': ['manzana', 'naranja', 'pera', 'mango'],
                    'precios': [1, 2, 3, 5]})
df2 = pd.DataFrame({'Frutas y Verduras': ['manzana', 'zanahoria', 'tomate', 'pera'],
                    'precios': [5, 6, 7, 8]})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1, df2, how = "inner", on = "precios")
```

```python
pd.merge(df1, df2, how = "left", on = "precios")
```

```python
pd.merge(df1, df2, how = "right", on = "precios")
```

```python
pd.merge(df1, df2, how = "outer", on = "precios")
```

```python
df_1 = pd.DataFrame({'Frutas': ['manzana', 'naranja', 'pera', 'mango'],
                     'unidades':[34, 82, 15, 20],
                    'id_vendedor': [1, 2, 3, 5]})
df_2 = pd.DataFrame({'Frutas y Verduras': ['manzana', 'zanahoria', 'tomate', 'pera'],
                     'unidades':[72, 41, 70, 2],
                    'id_vendedor': [5, 6, 7, 8]})
df_1
```

```python
df_2
```

```python
pd.merge(df_1, df_2, how = "left", on = "id_vendedor", suffixes = ("_tabla1", "_tabla2"))
```

```python
pd.concat([df1, df2], axis = 1)
```

```python
pd.concat([df1, df2], axis = 0)
```

```python
pd.concat([df1, df2], axis = 0, ignore_index = True)
```

```python
pd.concat([df1, df2], axis = 0, ignore_index = True)[["Frutas", "Frutas y Verduras"]]
```

### Tablas Dinámicas (Tablas Resumen) 

```python
import numpy as np
```

```python
titanic.pivot_table(index = "Sexo", columns = "Pclass", values = "Id_Pasajero", aggfunc = "nunique")
```

```python
titanic.pivot_table(index = ["Sexo", "Pclass"], columns = ["Embarked"], values = "Id_Pasajero", aggfunc = "nunique")
```

```python
titanic.pivot_table(index = ["Sexo", "Pclass"], columns = ["Embarked"], values = "Id_Pasajero", aggfunc = "nunique", 
                   fill_value = 0)
```

```python
titanic.pivot_table(index = ["Sexo", "Parch"], columns = ["Embarked"], values = "Edad", aggfunc = np.median, 
                   fill_value = 0.0)
```

```python
titanic.pivot_table(index = ["Embarked"], columns = ["Sexo", "Parch"], values = "Edad", aggfunc = np.median, 
                   fill_value = 0.0)
```
