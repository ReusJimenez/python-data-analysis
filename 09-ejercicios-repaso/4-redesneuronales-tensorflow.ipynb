### Ejemplos prácticos

#### Caso 1: Clasificación

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Flatten
from tensorflow import keras
import tensorflow as tf
from keras import activations
```

```python
clasificacion = pd.read_csv("ML-MATT-CompetitionQT2021_train.csv", sep = ";")
clasificacion.head()
```

```python
data = clasificacion[['PRBUsageUL', 'PRBUsageDL', 'meanThr_DL',
       'meanThr_UL', 'maxThr_DL', 'maxThr_UL', 'meanUE_DL', 'meanUE_UL',
       'maxUE_DL', 'maxUE_UL', 'maxUE_UL+DL', 'Unusual']]

x_data = data[['PRBUsageUL', 'PRBUsageDL', 'meanThr_DL',
       'meanThr_UL', 'maxThr_DL', 'maxThr_UL', 'meanUE_DL', 'meanUE_UL',
       'maxUE_DL', 'maxUE_UL', 'maxUE_UL+DL']]

y_data = data["Unusual"]

x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size = 0.2, random_state = 1000)
```

```python
model_c = Sequential()
model_c.add(Dense(units = 32, activation = "relu"))#, input_dim = len(X_train.columns)))
model_c.add(Dense(units = 64, activation = "relu"))
model_c.add(Dense(units = 1, activation = "sigmoid"))
```

```python
model_c.compile(loss = "binary_crossentropy", optimizer = tf.keras.optimizers.SGD(learning_rate = 0.1), metrics = ["accuracy"])
```

```python
model_c.summary()
```

```python
model_c.fit(x_train, y_train, epochs = 20, batch_size = 32) #200
```

```python
model_c.summary()
```

```python
y_pred_NN_c = model_c.predict(x_test)
y_pred_NN_c = np.round(y_pred_NN_c, 0)
y_pred_NN_c
```

```python
print(confusion_matrix(y_pred_NN_c, y_test))
print(accuracy_score(y_pred_NN_c, y_test))
print(precision_score(y_pred_NN_c, y_test))
print(recall_score(y_pred_NN_c, y_test))
print(f1_score(y_pred_NN_c, y_test))
print(roc_auc_score(y_pred_NN_c, y_test))
```

#### Caso 2: Regresión

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Flatten
from tensorflow import keras
import tensorflow as tf
from keras import activations
```

```python
regresion = pd.read_csv("House_Prices_train.csv")
regresion.head()
```

```python
nx_data = regresion[['Id', 'MSSubClass', 'LotArea', 'OverallQual',
       'OverallCond', 'YearBuilt', 'YearRemodAdd', 'BsmtFinSF1',
       'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', '1stFlrSF', '2ndFlrSF',
       'LowQualFinSF', 'GrLivArea', 'BsmtFullBath', 'BsmtHalfBath', 'FullBath',
       'HalfBath', 'BedroomAbvGr', 'KitchenAbvGr', 'TotRmsAbvGrd',
       'Fireplaces', 'GarageCars', 'GarageArea', 'WoodDeckSF',
       'OpenPorchSF', 'EnclosedPorch', '3SsnPorch', 'ScreenPorch', 'PoolArea',
       'MiscVal', 'MoSold', 'YrSold']]

ny_data = regresion["SalePrice"]

nx_data_ = nx_data.dropna()
ny_data_ = ny_data.dropna()

nx_train, nx_test, ny_train, ny_test = train_test_split(nx_data_, ny_data_, test_size = 0.2, random_state = 1000)
```

```python
model_r = Sequential()
model_r.add(Dense(units = 32, activation = "relu", input_dim = len(nx_train.columns)))
model_r.add(Dense(units = 64, activation = "relu"))
model_r.add(Dense(units = 1))
```

```python
model_r.summary()
```

```python
model_r.compile(loss = "mse", optimizer = "Adam", metrics = ['mae', 'mse']) #error absoluto medio (MAE)
```

```python
"keras.optimizers.SGD(learning_rate = 0.01)", "mean_squared_error"
"tf.keras.optimizers.Adam(learning_rate = 0.01)"
```

```python
model_r.summary()
```

```python
model_r.fit(nx_train, ny_train, epochs = 20, batch_size = 128)#32
```

```python
history = model_r.fit(nx_train, ny_train,
  epochs = 50, validation_split = 0.2, verbose=0)
history
```

```python
y_pred_NN_r = model_r.predict(nx_test)
y_pred_NN_r
```

```python
mean_squared_error(ny_test, y_pred_NN_r)
```

```python
# Graficacion

def plot_history(history):
    hist = pd.DataFrame(history.history)
    hist['epoch'] = history.epoch

    plt.figure()
    plt.xlabel('Epoch')
    plt.plot(hist['epoch'], hist['mae'], label='Train Error')
    plt.plot(hist['epoch'], hist['val_mae'], label = 'Val Error')
    plt.legend()

    plt.figure()
    plt.xlabel('Epoch')
    plt.plot(hist['epoch'], hist['mse'], label='Train Error')
    plt.plot(hist['epoch'], hist['val_mse'], label = 'Val Error')
    plt.legend()
    plt.show()

plot_history(history)
```
